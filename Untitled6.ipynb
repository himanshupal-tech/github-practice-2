```python
import random
impoimportrt time

words = [
    "python",
    "coding",
    "machine",
    "computer",
    "developer",
    "artificial",
    "intelligence",
    "program",
    "function",
    "variable"
]

random.shuffle(words)

score = 0
total_words = 5
time_limit = 15

print("=" * 35)
print("     TYPING SPEED CHALLENGE")
print("=" * 35)
print(f"\nYou have {time_limit} seconds to type {total_words} words!")
print("Type each word correctly to earn 20 points.")

input("\nPress Enter to start...")

start_time = time.time()

for word in words[:total_words]:
    elapsed = time.time() - start_time
    remaining = time_limit - elapsed

    if remaining <= 0:
        print("\nTime's up!")
        break

    print(f"\nTime left: {remaining:.1f} seconds")
    print("Type this word:", word)

    answer = input("Your answer: ")

    if time.time() - start_time >= time_limit:
        print("\nTime's up!")
        break

    if answer == word:
        print("Correct! +20 points")
        score += 20
    else:
        print("Incorrect!")

print("\n" + "=" * 35)
print("           GAME OVER")
print("=" * 35)
print(f"Final Score: {score}")
print(f"Correct Words: {score // 20}/{total_words}")
print("=" * 35)
```
